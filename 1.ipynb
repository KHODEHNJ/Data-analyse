```
import pandas as pd
f= pd.read_csv("/content/9. Sales-Data-Analysis.csv")
```

```
f
```

```
import matplotlib.pyplot as plt

f["Purchase Type"].value_counts().plot(
    kind="bar",
    figsize=(8, 5),
    color="steelblue"
)

plt.title("Number of Sales by Purchase Type")
plt.xlabel("Purchase Type")
plt.ylabel("Number of Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


counts = f["Purchase Type"].value_counts()

result = pd.DataFrame({
    "Count": counts,
    "Percentage": (counts / len(f) * 100).round(2)
})

display(result)
```

```
sales_by_product = f.groupby("Product")["Quantity"].sum().sort_values(ascending=False)

print(sales_by_product)
```

```
import matplotlib.pyplot as plt

sales_by_product.plot(kind="bar", figsize=(10, 5))

plt.title("Total Quantity Sold by Product")
plt.xlabel("Product")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```
sales = f.groupby("Product")["Quantity"].sum()

sales.plot(kind="bar" , figsize=(10,5))

plt.title("Total Quantity Sold by Product")
plt.xlabel("Product")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=45)
```
